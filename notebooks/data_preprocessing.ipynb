# Student Performance & Churn Analysis - Data Preprocessing

This notebook focuses on cleaning and transforming the raw university student database into a standardized format suitable for Exploratory Data Analysis (EDA) and Predictive Modeling.

## 1.0 Imports and Configuration

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set default visualization style
sns.set_theme(style="whitegrid")
```

## 2.0 Load and Cleaning

```python
# 1. Beolvasás (kezeljük a magyar tizedesvesszőt és szeparátort)
df_raw = pd.read_csv('../data/AA_mintafeladat.csv', sep=';', decimal=',', encoding='latin1')
df_raw.head()
```

## 3.0 Column Standardization

```python
COLUMN_TRANSLATION = {
    'Key': 'student_id',
    'Képzés jogviszony kezdete': 'enrollment_date',
    'Modulkód': 'major_code',
    'KépzésStátusz': 'status_at_record',
    'Tagozat': 'study_mode',
    'Képzési szint': 'degree_level',
    'Félév': 'semester_id',
    'Pénzügyi státusz': 'funding_type',
    'FélévesStátusz': 'semester_active_status',
    'Nem': 'gender',
    'Születési dátum': 'birth_date',
    'Irányítószám': 'zip_code',
    'FélévSorszám': 'term_number',
    'Aktív félévek': 'total_active_terms',
    'Féléven felvett kredit': 'credits_taken_term',
    'Féléven kumulált felvett kredit': 'credits_taken_cum',
    'Féléven teljesített kredit': 'credits_done_term',
    'Féléven kumulált teljesített kredit': 'credits_done_cum',
    'Átlag': 'gpa_term',
    'Kumulált átlag': 'gpa_cum'
}

df = df_raw.rename(columns=COLUMN_TRANSLATION)
df.head()
```

## 5.0 Temporal Feature Engineering

```python
temp_year = df['birth_date'].str.slice(-2).astype(int)
df['birth_date'] = temp_year.apply(lambda x: 1900 + x if x > 25 else 2000 + x)
df.rename(columns={'birth_date': 'birth_year'}, inplace=True)

df.head()
```

## 5.1 Age at Enrollment

```python
df['enrollment_year'] = pd.to_datetime(df['enrollment_date'], errors='coerce').dt.year
df['age_at_enrollment'] = df['enrollment_year'] - df['birth_year']
df['age_at_enrollment'] = df['age_at_enrollment'].clip(lower=18, upper=70)

print("Age at enrollment distribution:")
print(df['age_at_enrollment'].describe())
```

## 6.0 Find Missing Values

```python
# Hiányzó értékek abszolút és százalékos aránya
missing_data = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage (%)': (df.isnull().sum() / len(df)) * 100
}).sort_values(by='Percentage (%)', ascending=False)

print("--- MISSING DATA REPORT ---")
print(missing_data[missing_data['Missing Values'] > 0])
```

## 6.1 Handle Missing Values

## 8.0 Funding Type Simplification

```python
state_funded_keywords = [
    'State-funded', 'State-supported', 'State-scholarship', 
    'Ministerial scholarship', 'Partial state scholarship',
    'A1 category foreign scholarship', 'C category foreign scholarship',
    'B foreign external support scholarship'
]

df['funding_category'] = df['funding_type'].apply(
    lambda x: 'State-funded' if any(key in str(x) for key in state_funded_keywords) else 'Self-funded'
)

print(df['funding_category'].value_counts())
```

## 8.1 Churn Target Definition

```python
churn_statuses = [
    'Deleted', 
    'Left without notice', 
    'Failed final exam', 
    'Deleted upon own request', 
    'Inactive'
]

df['is_churned'] = df['status_at_record'].isin(churn_statuses).astype(int)

print("Churn distribution (Target Variable):")
print(df['is_churned'].value_counts(normalize=True) * 100)
```

## 9.2 Academic Efficiency

```python
df['academic_efficiency'] = np.where(
    df['credits_taken_cum'] > 0, 
    df['credits_done_cum'] / df['credits_taken_cum'], 
    0.0
)

df['academic_efficiency'] = df['academic_efficiency'].round(3)

print("Average academic efficiency by university:")
print(df.groupby('university')['academic_efficiency'].mean())
```

## 9.3 Major Field Categorization

```python
def get_broad_field(major_name):
    name = str(major_name)
    if any(x in name for x in ['Business', 'Finance', 'Accounting', 'Management', 'Economics', 'Marketing', 'Commerce']):
        return 'Economics & Business'
    elif 'Informatics' in name:
        return 'IT'
    elif any(x in name for x in ['Food', 'Landscape']):
        return 'Agriculture & Engineering'
    elif any(x in name for x in ['International', 'Communication', 'Sociology']):
        return 'Social Sciences'
    elif 'Tourism' in name:
        return 'Tourism'
    return 'Other'


df['major_field_broad'] = df['major_name_en'].apply(get_broad_field)

print("Records per Broad Field:")
print(df['major_field_broad'].value_counts())
```

## 9.4 Column Reordering (Logical Schema)

```python
ordered_columns = [
    # 1. Identifiers & Dates
    'student_id', 'semester_id', 'term_number', 'enrollment_date', 'enrollment_year',
    
    # 2. Demographics
    'gender', 'birth_year', 'age_at_enrollment',
    
    # 3. Geography
    'zip_code', 'county',
    
    # 4. Academic Background
    'university', 'major_code', 'major_name_en', 'major_field_broad', 
    'degree_level', 'study_mode',
    
    # 5. Financials & Status
    'funding_type', 'funding_category', 'semester_active_status', 'status_at_record',
    
    # 6. Performance Metrics
    'total_active_terms', 
    'credits_taken_term', 'credits_taken_cum', 
    'credits_done_term', 'credits_done_cum', 
    'gpa_term', 'gpa_cum', 'academic_efficiency',
    
    # 7. Target Variable
    'is_churned'
]

current_cols = set(df.columns)
ordered_set = set(ordered_columns)

missing_from_order = current_cols - ordered_set
if missing_from_order:
    print(f"Warning: These columns were not in the list: {missing_from_order}")
    ordered_columns += list(missing_from_order)

df = df[ordered_columns]

print("Columns successfully reordered. New schema preview:")
print(df.iloc[:, :29].info()) 
```
